````python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt

print("Python", sys.version.split()[0], "| pandas", pd.__version__)
print("环境 OK!开始今天的练习 →")

raw = pd.read_csv(os.path.join("..", "datasets", "titanic-sample.csv"))
````

## 六步模板(今天全程照做)

1. 概览:shape / dtypes / 缺失
2. 清洗:Age 中位数、Embarked 众数
3. 特征:FamilySize = SibSp + Parch + 1;IsChild = Age < 16
4. 单变量:总存活率、各特征分布
5. 双变量:分组交叉 + 相关性排序
6. 结论 + 下一步(下周 W3 进吴恩达 ML Specialization)


````python
def eda(df):
    print("STEP 1 概览")
    print("形状:", df.shape, "| 缺失:", df.isna().sum().sum())

    print("STEP 2 清洗")
    df = df.copy()
    df["Age"] = df["Age"].fillna(df["Age"].median())
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
    print("剩余缺失:", df.isna().sum().sum())

    print("STEP 3 造特征")
    df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
    df["IsChild"] = (df["Age"] < 16).astype(int)
    print(df[["Age", "SibSp", "Parch", "FamilySize", "IsChild"]].head())

    print("STEP 4 单变量")
    print("总存活率:", round(df["Survived"].mean() * 100, 1), "%")
    print("儿童占比:", round(df["IsChild"].mean() * 100, 1), "%")

    print("STEP 5 双变量")
    features = ["Pclass", "Age", "SibSp", "Parch", "Fare", "FamilySize", "IsChild"]
    corr = df[["Survived"] + features].corr()["Survived"].sort_values(ascending=False).round(3)
    print("与存活的相关性:\n", corr)
    return df

clean = eda(raw)
````

## 任务 2:两张收尾图

左图:FamilySize 分组的存活率;右图:Fare 按存活分组的箱线图。


````python
plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
clean.groupby("FamilySize")["Survived"].mean().plot(kind="bar", color="steelblue", rot=0)
plt.xlabel("Family size")
plt.ylabel("Survival rate")
plt.title("Survival by family size")

plt.subplot(1, 2, 2)
clean.boxplot(column="Fare", by="Survived", grid=False)
plt.suptitle("")
plt.xlabel("Survived")
plt.ylabel("Fare")
plt.title("Fare by survival")

plt.tight_layout()
plt.show()
````

````python
print("STEP 6 结论")
print("1. 女性、一二等舱、儿童存活率明显更高 —— 性别/舱位/年龄是可用的特征")
print("2. 相关性排序里 Pclass、Fare、Age 排前列,与直觉一致")
print("3. 下一步:W3 起进入吴恩达 ML Specialization,这些 EDA 习惯会直接用在建模前")
````
