## 任务 1:melt —— 宽表转长表

- 宽表:一列一个月(1月、2月、3月…),适合人看
- 长表:一行一个观测(产品, 月份, 销量),适合机器分析/绘图

`melt(id_vars="产品", var_name="月份", value_name="销量")`:
- `id_vars` → 保持不动的列
- `var_name` / `value_name` → 新生成的"变量名"和"值"列


````python
wide = pd.DataFrame({
    "产品": ["A", "B", "C"],
    "1月": [10, 20, 15],
    "2月": [12, 22, 18],
    "3月": [14, 25, 20],
})
print("宽表:\n", wide)

long = wide.melt(id_vars="产品", var_name="月份", value_name="销量")
print("长表:\n", long)
````

## 任务 2:pivot / pivot_table —— 长表转宽表

- `pivot(index="产品", columns="月份", values="销量")` → 还原宽表
- `pivot_table(..., aggfunc="sum")` → 有重复组合时先聚合再摊开(更稳)


````python
back = long.pivot(index="产品", columns="月份", values="销量")
print("pivot 还原:\n", back)

piv = pd.pivot_table(long, index="产品", columns="月份", values="销量", aggfunc="sum")
print("pivot_table:\n", piv)
````

## 任务 3:stack / unstack(多级索引)

`stack()` 把**列**压成行(宽→长),`unstack()` 把**行**摊成列(长→宽),互为逆操作。


````python
s = pd.Series([10, 20, 30, 40], index=[["A", "A", "B", "B"], ["x", "y", "x", "y"]])
print("多级索引 Series:\n", s)
print("unstack 摊成宽表:\n", s.unstack())
print("再 stack 回去:\n", s.unstack().stack())
````

## 任务 4:小挑战 🔥(成绩表三连)

1. 宽表 → 长表(melt)
2. 按科目算 mean / max(groupby)
3. 再 pivot 回宽表


````python
scores_wide = pd.DataFrame({
    "姓名": ["小明", "小红", "小刚"],
    "数学": [85, 92, 78],
    "英语": [88, 95, 72],
})
scores_long = scores_wide.melt(id_vars="姓名", var_name="科目", value_name="分数")
print("长表:\n", scores_long)

stats = scores_long.groupby("科目")["分数"].agg(["mean", "max"]).round(1)
print("按科统计:\n", stats)

back = scores_long.pivot(index="姓名", columns="科目", values="分数")
print("还原宽表:\n", back)
````
