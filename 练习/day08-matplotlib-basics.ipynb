## 任务 1:Figure 和 Axes(一张图 = 画布 + 坐标系)

- `fig, ax = plt.subplots()` → 一张画布(fig)+ 一个坐标系(ax)
- `ax.plot(x, y)` → 在坐标系里画线
- `ax.set_xlabel / set_ylabel / set_title` → 标签和标题
- `ax.grid(True)` → 网格


````python
x = np.linspace(0, 2 * np.pi, 100)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, np.sin(x), color="tab:blue", linewidth=2)
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.set_title("Sine wave")
ax.grid(True)
plt.show()
````

## 任务 2:四种基础图(先会选图,再会画)

| 图 | 函数 | 看什么 |
|---|---|---|
| 折线 | `ax.plot` | 趋势(随 x 变化) |
| 散点 | `ax.scatter` | 两个变量的关系 |
| 柱状 | `ax.bar` | 类别之间比较 |
| 直方 | `ax.hist` | 单变量的分布 |


````python
rng = np.random.default_rng(8)
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

x = np.linspace(0, 10, 50)
y = 2 * x + 1 + rng.normal(0, 2, 50)
axes[0, 0].plot(x, 2 * x + 1, "b-")
axes[0, 0].set_title("Line")
axes[0, 1].scatter(x, y, s=15, alpha=0.7)
axes[0, 1].set_title("Scatter")
axes[1, 0].bar(["A", "B", "C", "D"], [3, 7, 5, 9], color="orange")
axes[1, 0].set_title("Bar")
axes[1, 1].hist(rng.normal(0, 1, 500), bins=20, color="green")
axes[1, 1].set_title("Hist")

plt.tight_layout()
plt.show()
````

## 任务 3:子图布局与共享坐标轴

- `plt.subplots(2, 1)` → 上下两行
- `sharex=True` → 两图共用 x 轴(比较波动最好用)
- `plt.tight_layout()` → 自动收一收标签,防重叠


````python
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(6, 4))
x = np.linspace(0, 10, 100)
ax1.plot(x, np.sin(x))
ax1.set_title("sin")
ax2.plot(x, np.cos(x), color="r")
ax2.set_title("cos")
plt.tight_layout()
plt.show()
````

## 任务 4:图例、线型与保存

- `label="..."` + `ax.legend()` → 图例
- `linestyle` 取值 `-` / `--` / `:` 等
- `fig.savefig("名字.png", dpi=150)` → 存图(dpi 越高越清晰)


````python
fig, ax = plt.subplots(figsize=(6, 3))
x = np.linspace(0, 5, 100)
ax.plot(x, x, label="linear", linestyle="-")
ax.plot(x, x ** 2, label="squared", linestyle="--")
ax.plot(x, np.sqrt(x), label="sqrt", linestyle=":")
ax.set_xlabel("x")
ax.set_title("Line styles")
ax.legend()
ax.grid(True)
fig.savefig("day08-demo.png", dpi=150)
print("已保存: day08-demo.png(练习目录下)")
plt.show()
````

## 任务 5:小挑战 🔥(30 天气温)

造 30 天的模拟气温(正弦趋势 + 噪声),完成:
1. 画每日气温折线 + 7 日移动平均线
2. 用红点标出最高温那天,并加注释

提示:`np.convolve(temps, np.ones(7) / 7, mode="same")` 可以做移动平均。


````python
rng = np.random.default_rng(18)
days = np.arange(1, 31)
temps = 20 + 8 * np.sin(days / 7) + rng.normal(0, 2, 30)
smooth = np.convolve(temps, np.ones(7) / 7, mode="same")

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(days, temps, label="daily", alpha=0.6)
ax.plot(days, smooth, label="7-day avg", linewidth=2)
peak_day = days[temps.argmax()]
ax.scatter([peak_day], [temps.max()], color="red", zorder=5, label="peak")
ax.annotate(
    "peak",
    xy=(peak_day, temps.max()),
    xytext=(peak_day + 3, temps.max() - 1),
    arrowprops=dict(arrowstyle="->"),
)
ax.set_xlabel("Day")
ax.set_ylabel("Temperature")
ax.set_title("30 days temperature")
ax.legend()
plt.show()
````
