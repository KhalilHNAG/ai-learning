## 任务 1:Series —— 带标签的一维数组

NumPy 数组只有位置(0,1,2...),Series 多了一套**标签(index)**:

- `pd.Series([10, 20, 30], index=["a", "b", "c"])` → 用字母当下标
- `.values` → 底层数据;`.index` → 标签;`.dtype` → 类型
- 有歧义时明确表态:`.loc[]` 按标签、`.iloc[]` 按位置

先猜:`s["a"]` 和 `s.iloc[0]` 结果一样吗?


````python
s = pd.Series([10, 20, 30], index=["a", "b", "c"])
print(s)
print("values:", s.values)
print("index:", s.index.tolist())
print("按标签 s['a']:", s["a"])
print("按位置 s.iloc[0]:", s.iloc[0])
print("大于 15 的元素:\n", s[s > 15])
````

## 任务 2:DataFrame —— 二维表格

用 `pd.DataFrame(字典)` 建表:每个 key 是一列,各列长度必须一致。

常用动作:
- `df.shape` / `df.columns` / `df.dtypes` → 几行几列、列名、每列类型
- `df["新列"] = ...` → 加一列;`df.sort_values("列名")` → 排序
- `df.head(n)` → 看前 n 行


````python
df = pd.DataFrame({
    "姓名": ["小明", "小红", "小刚", "小丽"],
    "数学": [85, 92, 78, 96],
    "英语": [88, 95, 72, 90],
})
print("形状:", df.shape, "| 列:", df.columns.tolist())
print("每列类型:\n", df.dtypes)

df["总分"] = df["数学"] + df["英语"]
df["平均分"] = df[["数学", "英语"]].mean(axis=1).round(1)
print(df.sort_values("总分", ascending=False))
````

## 任务 3:读写 CSV

CSV 就是逗号分隔的纯文本表格,数据科学生态里最重要的交换格式。

- 写入:`df.to_csv(路径, index=False)`——`index=False` 要养成习惯(不把行号写进文件)
- 读回:`pd.read_csv(路径)`,再看 `.shape` / `.head()`
- 用相对路径 `../datasets/...` 写到仓库的 datasets 目录(从练习文件夹往上一级)


````python
import os

csv_path = os.path.join("..", "datasets", "day05-students.csv")
os.makedirs(os.path.dirname(csv_path), exist_ok=True)
df.to_csv(csv_path, index=False)
print("已写入:", os.path.abspath(csv_path))

df2 = pd.read_csv(csv_path)
print("读回形状:", df2.shape)
print(df2.head())
print("数学列平均:", df2["数学"].mean())
````

## 任务 4:缺失值第一眼

缺失值在 pandas 里统一是 `NaN`。三件套:

- `df.isna().sum()` → 每列有多少个缺失
- `df.dropna()` → 丢掉含缺失的行(返回新表)
- `df.fillna(值)` → 用指定值补上(数值列常用中位数)

pandas 3 默认 Copy-on-Write:修改用**赋值回变量**,别写 `inplace=True`。


````python
df3 = pd.DataFrame({
    "产品": ["A", "B", "C", "D", "E"],
    "销量": [100, None, 130, None, 90],
    "评分": [4.5, 3.8, None, 4.9, None],
})
print("缺失计数:\n", df3.isna().sum())

fixed = df3.copy()
fixed["销量"] = fixed["销量"].fillna(fixed["销量"].median())
print("补销量(中位数):\n", fixed)
print("丢掉缺评分行:\n", df3.dropna(subset=["评分"]))
````

## 任务 5:三张"速查表"

- `df.describe()` → 数值列的 count/mean/std/min/四分位/max,一眼看分布
- `df["类别列"].value_counts()` → 每个取值出现几次
- `df["列"].nunique()` → 去重后有几个不同取值


````python
print("数值列概况:\n", df.describe().round(1))

people = pd.DataFrame({"性别": ["男", "女", "男", "女", "男", "女", "男"]})
print("性别分布:\n", people["性别"].value_counts())
print("不同取值数:", people["性别"].nunique())
````
