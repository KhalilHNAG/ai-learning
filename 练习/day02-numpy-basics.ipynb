## 任务 2:查看数组的属性

数组(`ndarray`)自带四个常看属性:`shape`(形状)、`ndim`(维度数)、`size`(元素总数)、`dtype`(元素类型)。

对随机矩阵 `np.random.rand(3, 4)` 完成打印:

````python
m = np.random.rand(3, 4)   # 3行4列,元素是0~1随机小数
print("矩阵:\n", m)
print("shape:", m.shape)
print("ndim:", m.ndim)
print("size:", m.size)
print("dtype:", m.dtype)
````

## 任务 3:形状变换

把一维数组 `np.arange(12)` 变成 3×4 矩阵,再变回一维:
- `reshape(3, 4)`:重排形状(元素个数必须一致,12 = 3×4)
- `M.T`:转置(行变列)
- `flatten()`:摊平回一维

````python
v = np.arange(12)
M = v.reshape(3, 4)
print("reshape:", M.shape, "\n", M)
print("转置:\n", M.T)
print("flatten:", M.flatten())
````

## 任务 4:广播复习(昨天的坑)

`a` 形状 (2,3)。两个候选 `b`,哪个能直接相加?先心算再运行:
- `b1 = np.array([10, 20, 30])` → 形状 `(3,)`
- `b2 = np.array([[10], [20]])` → 形状 `(2,1)`

规则回忆:维度**从右往左**对齐,不相等时**其中一个必须是 1**。

````python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
b1 = np.array([10, 20, 30])   # 形状 (3,) → 和 a 的列对齐,行方向广播
b2 = np.array([[10], [20]])   # 形状 (2,1) → 和 a 的行对齐,列方向广播
print("a + b1(行广播):\n", a + b1)
print("a + b2(列广播):\n", a + b2)
````

## 任务 5:小挑战 🔥

生成 5×5 随机整数矩阵(0~99),然后完成四问:
1. 打印第 2 行(索引 1)
2. 打印第 3 列(索引 2)→ 用 `M[:, 2]`
3. 每行的和 → `M.sum(axis=1)`(提前认识 `axis`,以后天天见)
4. 全矩阵平均值 → `M.mean()`

````python
M = np.random.randint(0, 100, size=(5, 5))
print(M)
print("第2行:", M[1])
print("第3列:", M[:, 2])
print("每行和(axis=1):", M.sum(axis=1))
print("平均值:", M.mean())
````
