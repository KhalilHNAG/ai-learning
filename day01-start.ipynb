## 练习 1:NumPy 数组与广播

````python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
print("形状:", a.shape)
print("转置:\n", a.T)

# 广播规则:维度从右往左对齐,不相等时其中一个必须是 1
b = np.array([[1, 1, 1]])        # 形状 (1,3) → 广播成 (2,3)
print("b 的形状:", b.shape)
print("广播加法:\n", a + b)

# 思考:如果把 b 改成 np.array([[1], [1], [1]])(形状 (3,1)),
# 与 (2,3) 相加会报错——为什么?取消下面一行注释亲自试一下:
# c = np.array([[1], [1], [1]]); print(a + c)
````

## 练习 2:Pandas 表格与计算新列

````python
df = pd.DataFrame({
    "名字": ["小明", "小红", "小刚"],
    "身高_cm": [175, 162, 180],
    "体重_kg": [70, 50, 75],
})
df["BMI"] = df["体重_kg"] / (df["身高_cm"] / 100) ** 2
df
````
